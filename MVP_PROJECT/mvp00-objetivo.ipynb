# Objetivo

## Problema
Entender como o metagame competitivo de Magic: The Gathering se comportou nos torneios do mtgo.com entre 2022 e meados de 2024. Procurando saber quais cartas dominaram, como jogadores e decks se posicionaram estrategicamente (mainboard vs sideboard), e se existe relação entre o perfil de custo de mana de um deck e seu sucesso competitivo.

## Perguntas de negócio
1) Quais cartas (excluindo terrenos básicos) mais apareceram em decks ao longo do período, por formato, e essa composição do meta mudou entre 2022 e 2024?
2) Quais cartas funcionam como "staples" (presença constante no mainboard) versus "respostas situacionais" (predominantemente jogadas no sideboard)?
3) Decks com custo de mana médio (CMC) mais baixo têm maior taxa de sucesso em torneios eliminatórios do que decks de CMC mais alto?
4) Jogadores com maior volume de participação em ligas mantêm uma taxa de vitória consistente ao longo do tempo?

## Nota sobre terminologia: deck e CMC
Um deck é o baralho de cartas que um jogador monta e usa para jogar em um torneio, no caso de Magic: The Gathering, composto por um mainboard (a lista principal de cartas usada no início da partida) e um sideboard (até 15 cartas reservas, que podem ser usadas entre partidas de um mesmo confronto para responder a estratégias específicas do oponente).

CMC (Converted Mana Cost, ou Custo de Mana Convertido) é a soma total de mana necessária para jogar uma carta, resumida em um único número, independentemente da cor. Por exemplo, uma carta que custa `{2}{U}{U}` (2 de mana genérica + 2 de mana azul) tem CMC 4. É essa métrica que permite comparar o "custo" de cartas diferentes de forma padronizada, e assim, calcular o CMC médio de um deck inteiro.
