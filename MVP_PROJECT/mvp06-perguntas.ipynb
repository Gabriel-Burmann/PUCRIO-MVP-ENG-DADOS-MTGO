## Pergunta 2: Staples vs respostas situacionais (mainboard vs sideboard)
Cartas com alta presença total, comparando proporção jogada no sideboard.
Filtro de volume mínimo (50 aparições) para evitar ruído estatístico.

```python
%sql
SELECT
    c.nome_carta,
    c.type_line,
    COUNT(*) AS total_aparicoes,
    SUM(CASE WHEN fc.is_sideboard THEN 1 ELSE 0 END) AS aparicoes_sideboard,
    ROUND(SUM(CASE WHEN fc.is_sideboard THEN 1 ELSE 0 END) / COUNT(*), 3) AS proporcao_sideboard
FROM workspace.gold.fato_deck_carta fc
JOIN workspace.gold.dim_carta c ON fc.carta_id = c.carta_id
GROUP BY c.nome_carta, c.type_line
HAVING COUNT(*) >= 50
ORDER BY proporcao_sideboard DESC
LIMIT 20
```

```python
%sql
SELECT
    c.nome_carta,
    c.type_line,
    COUNT(*) AS total_aparicoes,
    SUM(CASE WHEN fc.is_sideboard THEN 1 ELSE 0 END) AS aparicoes_sideboard,
    ROUND(SUM(CASE WHEN fc.is_sideboard THEN 1 ELSE 0 END) / COUNT(*), 3) AS proporcao_sideboard
FROM workspace.gold.fato_deck_carta fc
JOIN workspace.gold.dim_carta c ON fc.carta_id = c.carta_id
GROUP BY c.nome_carta, c.type_line
HAVING COUNT(*) >= 50
ORDER BY proporcao_sideboard ASC
LIMIT 20
```

## Pergunta 3: CMC de decks vencedores ao longo do tempo
CMC médio (ponderado por quantidade) do mainboard, comparando campeões
(colocacao = 1) vs demais, por ano. Terrenos têm cmc = 0 e entram no
cálculo naturalmente (refletem o mana curve real do deck).

```python
%sql
-- Pauper
WITH cmc_por_deck AS (
    SELECT
        fc.deck_id,
        SUM(c.cmc * fc.quantidade) / SUM(fc.quantidade) AS cmc_medio
    FROM workspace.gold.fato_deck_carta fc
    JOIN workspace.gold.dim_carta c ON fc.carta_id = c.carta_id
    WHERE fc.is_sideboard = false
    GROUP BY fc.deck_id
)
SELECT
    YEAR(t.data) AS ano,
    CASE WHEN fr.colocacao = 1 THEN 'campeao' ELSE 'demais' END AS grupo,
    ROUND(AVG(cd.cmc_medio), 2) AS cmc_medio_grupo,
    COUNT(*) AS num_decks
FROM workspace.gold.fato_resultado_deck fr
JOIN workspace.gold.dim_deck d ON fr.deck_id = d.deck_id
JOIN workspace.gold.dim_torneio t ON fr.torneio_id = t.torneio_id
JOIN cmc_por_deck cd ON fr.deck_id = cd.deck_id
WHERE fr.tipo_resultado = 'colocacao'
  AND t.nome LIKE 'Pauper%'
GROUP BY YEAR(t.data), CASE WHEN fr.colocacao = 1 THEN 'campeao' ELSE 'demais' END
ORDER BY ano, grupo
```

## Pergunta 4: Consistência de jogadores frequentes
Taxa de vitória média em ligas, comparando jogadores com muitas participações
vs poucas.

```python
%sql
SELECT
    formato,
    `2022`, `2023`, `2024`,
    COALESCE(`2022`,0) + COALESCE(`2023`,0) + COALESCE(`2024`,0) AS Total
FROM (
    SELECT
        CASE
            WHEN nome LIKE 'Pauper%' THEN 'Pauper'
            WHEN nome LIKE 'Modern%' THEN 'Modern'
            WHEN nome LIKE 'Legacy%' THEN 'Legacy'
            WHEN nome LIKE 'Vintage%' THEN 'Vintage'
            WHEN nome LIKE 'Standard%' THEN 'Standard'
            WHEN nome LIKE 'Pioneer%' THEN 'Pioneer'
            WHEN nome LIKE 'Limited%' THEN 'Limited'
            ELSE 'Outro'
        END AS formato,
        YEAR(data) AS ano
    FROM workspace.gold.dim_torneio
)
PIVOT (
    COUNT(*) FOR ano IN (2022, 2023, 2024)
)
ORDER BY Total DESC
```

```python
%sql
SELECT nome, COUNT(*) AS total
FROM workspace.gold.dim_torneio
WHERE nome NOT LIKE 'Pauper%' AND nome NOT LIKE 'Modern%' AND nome NOT LIKE 'Legacy%'
  AND nome NOT LIKE 'Vintage%' AND nome NOT LIKE 'Standard%' AND nome NOT LIKE 'Pioneer%'
  AND nome NOT LIKE 'Limited%'
GROUP BY nome
ORDER BY total DESC
```

```python
%sql
WITH desempenho_jogador AS (
    SELECT
        j.jogador_id,
        COUNT(*) AS num_participacoes,
        SUM(fr.vitorias) AS total_vitorias,
        SUM(fr.vitorias + fr.derrotas) AS total_partidas,
        SUM(fr.vitorias) / SUM(fr.vitorias + fr.derrotas) AS taxa_vitoria
    FROM workspace.gold.fato_resultado_deck fr
    JOIN workspace.gold.dim_deck d ON fr.deck_id = d.deck_id
    JOIN workspace.gold.dim_jogador j ON d.jogador_id = j.jogador_id
    WHERE fr.tipo_resultado = 'record_liga'
    GROUP BY j.jogador_id
)
SELECT
    CASE WHEN num_participacoes >= 10 THEN 'frequente (10+)' ELSE 'ocasional (<10)' END AS perfil_jogador,
    ROUND(AVG(taxa_vitoria), 3) AS media,
    ROUND(STDDEV(taxa_vitoria), 3) AS desvio_padrao,
    ROUND(MIN(taxa_vitoria), 3) AS minimo,
    ROUND(MAX(taxa_vitoria), 3) AS maximo
FROM desempenho_jogador
GROUP BY CASE WHEN num_participacoes >= 10 THEN 'frequente (10+)' ELSE 'ocasional (<10)' END
```
