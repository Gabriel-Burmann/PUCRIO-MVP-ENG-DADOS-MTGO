# Considerações Finais

## Dificuldades encontradas
Ao longo da iteração do modelo, surgiram bugs sutis de qualidade de dados que não eram óbvios à primeira vista: torneios duplicados por republicação, colisão de `AnchorUri` como chave de deck, e a diferença de estrutura do Scryfall entre cartas partidas e cartas de dupla face. Houve ainda a necessidade de revisitar e corrigir decisões de modelagem já tomadas. Por exemplo, reconsiderar a deduplicação de decks após avaliar o risco de perda de dado legítimo.

## Trabalhos futuros
Entre as possibilidades de continuidade estão inferir arquétipo de deck a partir da composição de cartas (heurística ou clustering), permitindo perguntas sobre diversidade estratégica do meta; estender o recorte temporal, e investigar os 13 torneios classificados como "Outro" para refinar a extração de formato.
